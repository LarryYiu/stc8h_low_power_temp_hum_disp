```python
import pandas as pd
import numpy as np
import re
```

```python
df = pd.read_csv('segTable.csv')
df.columns = [int(re.search(r"\((\d+)\)", col).group(1)) if re.search(r"\((\d+)\)", col) else -1 for col in df.columns]
df[-1] = [int(re.search(r"\((\d+)\)", com).group(1)) if re.search(r"\((\d+)\)", com) else -1 for com in df[-1]]
df.set_index(-1, inplace=True)
df.index.name = None
```

```python
df
```

```python
encoded = df.copy()
for i in encoded.columns:
    for j in encoded.index:
        encoded.at[j, i] = f"#define __{encoded.at[j, i]}_BIT__ {int(i%8)}"
print(encoded)
```

```python
print("\n".join(encoded.values.flatten().tolist()))
```

```python
__LCD_REG_ADDR_LOOKUP = np.array([[0x90, 0x91, 0x92, 0x93, 0x94], [0x98, 0x99,
                                   0x9A, 0x9B, 0x9C], [0xA0, 0xA1, 0xA2, 0xA3,
                                   0xA4], [0xA8, 0xA9, 0xAA, 0xAB, 0xAC]])
```

```python
encoded = df.copy()
for i in encoded.columns:
    for j in encoded.index:
        encoded.at[j, i] = f"#define __{encoded.at[j, i]}_ADDR__ 0x{__LCD_REG_ADDR_LOOKUP[j, int(i/8)]:02X}"
print(encoded)
```

```python
print("\n".join(encoded.values.flatten().tolist()))
```
